```python
# 4. Inspect rows containing missing data
missing_data_rows = df[df.isnull().any(axis = 1)]
missing_data_rows.head(10)
```

```python
# 5. Drop missing data
df_dropped = df.dropna() # We removed rows containing missing values from the dataset using dropna
```

```python
df_dropped.shape
```

```python
# 6. Impute numerical missing values with the mean
df_mean = df.copy()
numerical_columns = df_mean.select_dtypes(include = ["int64", "float64"]).columns
numerical_columns
```

```python
for col in numerical_columns:
  df_mean[col] = df_mean[col].fillna(df_mean[col].mean())
```

```python
df_mean[numerical_columns].isnull().sum()
```

```python
# 7. Impute numerical missing values with the median
df_median = df.copy()
for col in numerical_columns:
  df_median[col] = df_median[col].fillna(df_median[col].median())
df_median[numerical_columns].isnull().sum()
```

```python
# 8. Impute numerical missing values with the mode
df_mode_numerical = df.copy()
for col in numerical_columns:
  mode_value = df_mode_numerical[col].mode()[0]
  df_mode_numerical[col] = df_mode_numerical[col].fillna(mode_value)
df_mode_numerical[numerical_columns].isnull().sum()
```

```python
# 9. Impute categorical missing values with the mode
df_mod_categoryk = df.copy()
categoryk_sutunlar = df_mod_categoryk.select_dtypes(include = ["object"]).columns
categoryk_sutunlar
```

```python
for col in categoryk_sutunlar:
  mode_value = df_mod_categoryk[col].mode()[0]
  df_mod_categoryk[col] = df_mod_categoryk[col].fillna(mode_value)
```

```python
df_mod_categoryk[categoryk_sutunlar].isnull().sum()
```

```python
# 10. Impute categorical missing values with a constant value
df_sabit_categoryk = df.copy()

for col in categoryk_sutunlar:
  df_sabit_categoryk[col] = df_sabit_categoryk[col].fillna("Unknown")
```

```python
df_sabit_categoryk[categoryk_sutunlar].isnull().sum()
```

```python
# 11. Updated dataset to be used in training
# - Impute numerical missing values with the median
# - Impute categorical missing values with the mode
```

```python
df_updated = df.copy()
```

```python
for col in numerical_columns:
  df_updated[col] = df_updated[col].fillna(df_updated[col].median())
```

```python
for col in categoryk_sutunlar:
  mode_value = df_updated[col].mode()[0]
  df_updated[col] = df_updated[col].fillna(mode_value)
```

```python
# Helper function: Normalize text strings
# Convert case, strip whitespace, handle Turkish character differences
def normalize_text(series):

  return(
      series.astype(str)
      .str.strip()
      .str.lower()
      .str.replace("i̇", "i", regex=False)
      .str.replace("ı", "i", regex=False)
      .str.replace("ö", "o", regex=False)
      .str.replace("ü", "u", regex=False)
      .str.replace("ç", "c", regex=False)
      .str.replace("ş", "s", regex=False)
      .str.replace("ğ", "g", regex=False)
  )
```

## Outliers
Amaç:
  1. Veri setindeki aykırı değerlerin tespit edilmesi ve incelenmesi
  2. Aykırı değerleri temizleyerek güncel veri setinin oluşturulması

Adımlar:
  1. Veri setini yükle
  2. Sayısal sütunları belirle
  3. Sayısal sütunların özet istatistiklerini hesapla
  4. IQR ile aykırı değer sınırlarını hesapla
  5. Aykırı değer içeren kayıtları tespit etmek
  6. Aykırı değer sayılarını sütun bazında incelemek
  7. Aykırı değer içeren kayıtları veri setinden temizlemek
  8. Temizlenmiş veri setini kaydetmek

  

```python
import pandas as pd
```

```python
# 1. Load the dataset
df = pd.read_csv("e_commerce_dataset_duplicates_removed.csv")
```

```python
# 2. Identify numerical columns
numerical_columns = df.select_dtypes(include = ["int64", "float64"]).columns.tolist()
numerical_columns
```

```python
# 3. Inspect summary statistics for numerical columns
df[numerical_columns].describe()
```

```python
"""unit_price
  IQR = Q3(75%) - Q1(25%)
      = 817 - 198 = 619
  Upper limit = Q3 + k*IQR
  Lower limit = Q1 - k*IQR
    k = 1 olsun
    Upper limit = 817 + 619 = 1436
    Lower limit = 198 - 619 = -421
  Outlier = samples falling outside the upper and lower limits
"""
```

```python
# 4. Calculate outlier limits using IQR
outlier_limits = {}

for col in numerical_columns:

  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)

  iqr = q3 - q1
  lower_limit = q1 - 1.5 * iqr
  upper_limit = q3 + 1.5 * iqr

  outlier_limits[col] = {
      "q1": q1,
      "q3": q3,
      "iqr": iqr,
      "lower_limit": lower_limit,
      "upper_limit": upper_limit
  }
```

```python
outlier_limits
```

```python
# 5. Detect records containing outliers
outlier_mask = pd.Series(False, index = df.index)

for col in numerical_columns:
  lower_limit = outlier_limits[col]["lower_limit"]
  upper_limit = outlier_limits[col]["upper_limit"]

  column_mask = (df[col] < lower_limit) | (df[col] > upper_limit)
  outlier_mask = outlier_mask | column_mask
```

```python
outlier_records = df[outlier_mask]
outlier_records
```

```python
# 6. Examine outlier counts on a column basis
for col in numerical_columns:

  lower_limit = outlier_limits[col]["lower_limit"]
  upper_limit = outlier_limits[col]["upper_limit"]

  outlier_count = ((df[col] < lower_limit) | (df[col] > upper_limit)).sum()
  print(f"{col}: {outlier_count}")
```

```python
# 7. Clean records containing outliers from the dataset
df_cleaned = df[~outlier_mask].copy()
df_cleaned.shape
```

```python
# 8. Save the cleaned dataset
df_cleaned.to_csv("e_commerce_dataset_outliers_handled.csv")
```

```python
# 7. Group-based analysis with multiple columns
category_musteri_tipi_analizi = df.groupby(["category", "customer_type"]).agg(
    {
        "total_amount": ["mean", "sum"],
        "customer_rating": "mean",
        "delivery_days": "mean"
    }
).round(1)
category_musteri_tipi_analizi
```

## Pivot Table Analysis

Adımlar:

  3. Şehir ve ödeme türüne göre pivot tablo analizi

```python
import pandas as pd
```

## Data Visualization

Amaç:
  1. Veri setindeki farklı değişkenleri uygun grafik türüyle görselleştirmek
  2. Grafikler üzerinden zaman değişimi, ilişki, karşılaştırma ve dağılım yapısını incelemek

Adımlar:
  1. Veri setini yüklemek
  2. Line plot ile zaman içindeki değişimi görselleştirmek
  3. Scatter plot ile iki sayısal değişken arasındaki ilişkiyi görselleştirmek
  4. Bar plot ile karşılaştırmalı görselleştirme yapmak
  5. Histogram ile sayısal dağılım görselleştirmek
  6. Box plot ile dağılım ve aykırı değer tespiti

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme()
```

```python
# 1. Load the dataset
df = pd.read_csv("e_commerce_dataset_outliers_handled.csv")
df.head()
```

```python
# 2. Visualize change over time with a line plot
# Convert order date to datetime
df["order_date"] = pd.to_datetime(df["order_date"])
df["order_month"] = df["order_date"].dt.month
df.head()
```

```python
monthly_total_amount = df.groupby("order_month")["total_amount"].sum().reset_index()
monthly_total_amount
```

```python
plt.figure(figsize = (6,3))
plt.plot(monthly_total_amount["order_month"], monthly_total_amount["total_amount"], marker = "o") # line plot
plt.title("Total Amount by Month")
plt.xlabel("Months")
plt.ylabel("Total Amount")
plt.xticks(monthly_total_amount["order_month"])
plt.tight_layout()
plt.show()
```

```python
# 3. Visualize the relationship between two numerical variables with a scatter plot
plt.figure(figsize = (6,4))
plt.scatter(df["unit_price"], df["total_amount"], alpha = 0.6, color = "red", s = 5)
plt.title("Relationship Between Unit Price and Total Amount")
plt.xlabel("Unit Price")
plt.ylabel("Total Amount")
plt.show()
```

```python
# 4. Create comparative visualizations with a bar plot
# Sum total amount by category and visualize with a bar plot
category_toplam_tutar = (
    df.groupby("category")["total_amount"]
    .sum()
    .sort_values(ascending = False)
    .reset_index()
)
category_toplam_tutar
```

```python
plt.figure(figsize = (8,4))
sns.barplot(data = category_toplam_tutar, x = "category", y = "total_amount")
plt.title("Total Amount by Category")
plt.xlabel("category")
plt.ylabel("total amount")
plt.xticks(rotation = 45)
plt.show()
```

```python
# 5. Visualize numerical distribution with a histogram
plt.figure(figsize = (6,5))
plt.hist(df["total_amount"], bins = 150)
plt.title("Total Amount Distribution")
plt.xlabel("total amount")
plt.ylabel("frequency")
plt.show()
```

```python
# 6. Detect distribution and outliers with a box plot
plt.figure(figsize = (10,5))
sns.boxplot(data = df, x = "category", y = "total_amount")
plt.xlabel("category")
plt.ylabel("total amount")
plt.title("Total Amount by Category dağılımı")
plt.show()
```

## Time Series Analysis

Amaç:
  1. Veri setindeki tarih bilgisini kullanarak zaman bazlı analiz yapmak

Adımlar:
  1. Veri setini yüklemek
  2. Sipariş tarihini datetime tipine dönüştür
  3. Tarih sütunundan zaman özellikleri türet
  4. Monthsa göre sipariş yoğunluğunu incele
  5. Ayın ilk yarısı ve ikinci yarısında ki harcama farkını incele
  6. Hafta içi ve hafta sonu sipariş davranışlarını karşılaştırma
  7. İki haftalık dönemlerde ortalama sipariş tutarı inceleme

  


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. veri setinin yüklenmesi
df = pd.read_csv("e_commerce_dataset_outliers_handled.csv")
df.head()
```

```python
# 2. Convert order date to datetime type
df["order_date"] = pd.to_datetime(df["order_date"])
df["order_date"].dtype
```

```python
# 3. Derive time features from the date column
df["order_year"] = df["order_date"].dt.year
df["order_month"] = df["order_date"].dt.month
df["order_day"] = df["order_date"].dt.day
df["day_of_week"] = df["order_date"].dt.day_name()
df["week_number"] = df["order_date"].dt.isocalendar().week.astype(int)
df.head()
```

```python
# 4. Order density by month
monthly_density = (
    df.groupby("order_month")
    .agg(
        order_count = ("order_id", "count"),
        unique_customer_count = ("customer_id", "nunique")
    )
    .reset_index()
    .sort_values("order_month")
)
monthly_density
```

```python
plt.figure(figsize = (8,3))
plt.plot(monthly_density["order_month"], monthly_density["order_count"], marker = "o", color = "red", label = "order_count")
plt.plot(monthly_density["order_month"], monthly_density["unique_customer_count"], marker = "o", color = "green", label = "unique_customer_count")
plt.title("Monthsa göre sipariş yoğunluğu ve benzersiz müşteri sayısı")
plt.ylabel("Order Count and Unique Customer Count")
plt.xlabel("Months")
plt.xticks(monthly_density["order_month"])
plt.legend()
plt.show()
```

```python
# 5. Examine spending difference between first and second half of the month
# Classify as first half vs second half of the month
df["month_period"] = df["order_day"].apply(
    lambda x: "First half of the month" if x <= 15 else "Second half of the month"
)
df.head()
```

```python
month_period_analysis = (
    df.groupby("month_period")
    .agg(
        total_amount = ("total_amount", "sum")
    )
    .round(1)
    .reset_index()
)
month_period_analysis
```

```python
plt.figure()
sns.barplot(data = month_period_analysis, x = "month_period", y= "total_amount")
plt.title("Ayın ilk yarısı ve ikinci yarısına göre total amount")
plt.xlabel("Month period")
plt.ylabel("Total amount")
plt.show()
```

```python
# 6. Compare weekday vs weekend order behaviors
df["week_type"] = df["order_date"].dt.dayofweek.apply(
    lambda x: "weekend" if x >=5 else "weekday"
)
df.head()
```

```python
week_type_analysis = (
    df.groupby("week_type")
    .agg(
        average_order_amount = ("total_amount", "mean"),
    )
    .round(1)
    .reset_index()
)
week_type_analysis
```

```python
plt.figure()
sns.barplot(data = week_type_analysis, x = "week_type", y = "average_order_amount")
plt.title("Average Order Amount: Weekdays vs Weekends")
plt.xlabel("Week type")
plt.ylabel("Average order amount")
plt.show()
```

```python
# 7. Average order amount in two-week periods
df["two_week_period"] = ((df["week_number"] - 1)//2) + 1
df.head()
```

```python
two_week_analysis = (
    df.groupby("two_week_period")
    .agg(
        average_order_amount = ("total_amount", "mean")
    )
    .round(2)
    .reset_index()
    .sort_values("two_week_period")
)
two_week_analysis
```

```python
plt.figure(figsize = (10,4))
plt.plot(two_week_analysis["two_week_period"], two_week_analysis["average_order_amount"], marker = "o")
plt.title("Average Order Amount by Two-Week Periods")
plt.xlabel("two-week period")
plt.xticks(two_week_analysis["two_week_period"])
plt.ylabel("ortalama sipariş tutarı")
plt.show()
```

## Data Storytelling and Business Insights

Amaç:
  1. Gerçek hayata benzeyen iş sorularını veri ile cevaplamak
  2. Analiz sonuçlarını tablo, grafik ve kısa iş yorumları ile sunmak

Adımlar:
  1. Hangi category hem yüksek gelir üretiyor hem de müşteri memnuniyeti açısından güçlü görünüyor?
  2. Hangi ödeme türleri daha yüksek ortalama sipariş tutarı üretiyor ve müşteri puanı açısından performans gösteriyor?
  3. Hangi şehir-category kombinasyonları en güçlü fırsatı sunuyor?
  

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("e_commerce_dataset_outliers_handled.csv")
df.head()
```

```python
# 1. hangi category hem yüksek gelir hem de müşteri memnuniyetini yüksek sağlıyor
category_raporu = (
    df.groupby("category")
    .agg(
        total_revenue = ("total_amount", "sum"),
        average_customer_rating = ("customer_rating", "mean")
    )
    .round(2)
)
```

```python
category_raporu = category_raporu.sort_values(by = ["total_revenue"], ascending=[False])
category_raporu
```

```python
# categoryye göre toplam gelir
plt.figure(figsize = (8,4))
sns.barplot(data = category_raporu, x = "category", y = "total_revenue")
plt.title("Kategoriye göre toplam gelir")
plt.xlabel("Kategori")
plt.ylabel("Toplam Gelir")
plt.show()
```

```python
# categoryye göre ortalama musteri puani
plt.figure(figsize = (8,4))
sns.barplot(data = category_raporu, x = "category", y = "average_customer_rating")
plt.title("Average Customer Rating by Category")
plt.xlabel("Kategori")
plt.ylabel("Customer Rating")
plt.show()
```

```python
# 2. Which payment methods generate higher average order amounts and how do they perform in customer ratings?
payment_method_report = (
    df.groupby("payment_method")
    .agg(
        average_order_amount = ("total_amount", "mean"),
        average_customer_rating = ("customer_rating", "mean")
    )
    .round(2)
    .sort_values("average_order_amount", ascending = False)
)
payment_method_report
```

```python
# ödeme türüne göre ortalama sipariş tutarı
plt.figure(figsize = (8,4))
sns.barplot(data = payment_method_report, x = "payment_method", y = "average_order_amount")
plt.title("Average Order Amount by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Average Order Amount")
plt.show()
```

```python
# Average customer rating by payment method
plt.figure(figsize = (8,4))
sns.barplot(data = payment_method_report, x = "payment_method", y = "average_customer_rating")
plt.title("Average Customer Rating by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Average customer rating")
plt.show()
```
